```python
import numpy as np
```

```python
class DiscreteSignal:

    def __init__(self, start_time, end_time):
        self.start_time = start_time
        self.end_time = end_time

        self.time_range = range(start_time, end_time + 1)
        self.signal_values = np.zeros(len(self.time_range))

    
    def set_value_at_time(self, t, value):
        if t in self.time_range:
            index = t - self.start_time
            self.signal_values[index] = value

    def get_value_at_time(self , t):
        if t in self.time_range:
            index = t - self.start_time
            return self.signal_values[index]
        
        return 0
    
    def shift(self, k):

        new_start = self.start_time + k
        new_end = self.end_time + k
        new_signal = DiscreteSignal(new_start, new_end)
        
        new_signal.signal_values = np.copy(self.signal_values)
        
        return new_signal
```

```python
class LtiSystem:
    def __init__(self, impulse_signal : DiscreteSignal):
        self.impulse_response = impulse_signal

    def output(self, input : DiscreteSignal):
        output_start = self.impulse_response.start_time + input.start_time
        output_end = self.impulse_response.end_time + input.end_time

        y = DiscreteSignal(output_start,output_end)

        for n in y.time_range:
            y_n = 0

            for k in input.time_range:
                x_k = input.get_value_at_time(k)
                h_nk = self.impulse_response.get_value_at_time(n-k)

                y_n += x_k * h_nk


            y.set_value_at_time(n,y_n)


        return y 

    
```

```python
x = DiscreteSignal(0, 2)
x.set_value_at_time(0, 1)
x.set_value_at_time(1, 2)
x.set_value_at_time(2, 1)

h = DiscreteSignal(0, 1)
h.set_value_at_time(0, 1)
h.set_value_at_time(1, -1)

system = LtiSystem(h)
y = system.output(x)

print("Output time range:", list(y.time_range))
print("Output signal values:", y.signal_values)
```
