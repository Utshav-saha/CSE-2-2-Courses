```python
arr6 = np.array([1,2,3,4], dtype = 'S')
print(arr6)
```

```python
arr6 = arr6.astype(float)
print(arr6)
```

```python
arr7 = arr6.copy()
```

```python
arr8 = arr6.view()
arr8[0] = 123
print(arr6[0])
```

```python
print(arr8.base)
print(arr7.base)
```

```python
arr9 = np.vstack((arr6, arr7))
print(arr9)
```

```python
arr10 = (arr9 >= 2) & (arr9 < 3)
print(arr10)

values, counts = np.unique(arr10, return_counts=True)
print(dict(zip(values, counts)))

np.count_nonzero(arr10)
print(arr10.sum())
```
