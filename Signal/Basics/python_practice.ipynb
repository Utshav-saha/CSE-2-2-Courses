```python
# Problem 1

x = int(input())
data = []
records = {}
for i in range(x):
    name , severity , treatment_time = input().split()
    severity = int(severity)
    treatment_time = int(treatment_time)
    data.append((name, severity, treatment_time))


for x in data:
    name, severity, treatment_time = x
    if name not in records:
        records[name] = [0,0]
    
    records[name][0] += treatment_time
    records[name][1] += severity 

for key in sorted(records):
    total_time, severity_score = records[key]
    print(key, total_time, severity_score)


# So for each key l in records, the lambda computes:
# (records[l][1], l)
# records[l][1] → the severity score for that name (since records[name] = [total_time, severity_score])
# l → the name itself, used as a tiebreaker

max_label = max(records, key=lambda l: (records[l][1], l))
print(f"TOP {max_label} {records[max_label][1]}")




```

```python
# Problem 3

class Employee:

    company_name = "abc corp"

    def __init__(self,name,emp_id, salary_list):
        self.name = name
        self.id = emp_id
        self.salary_list = salary_list
    
    def add_salary(self, salary):
        self.salary_list.append(salary)
    
    def average_salary(self):
        return sum(self.salary_list)/len(self.salary_list)
    
    def max_salary(self):
        return max(self.salary_list)
    
    def annual_income(self):
        return sum(self.salary_list)
    
    def __str__(self):
        return f"Employee Name: {self.name}\nEmployee ID: {self.id}\nAverage Salary: {self.average_salary()}\nMax Salary: {self.max_salary()}\nAnnual Income: {self.annual_income()}"
    

e1 = Employee("John", 123, [1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000, 10000, 11000, 12000])
print(e1)


```

```python
# Problem 2

import numpy as np

s = int(input())
m = int(input())

matrix = np.zeros((s, m), dtype=int)

for x in range(s):
    for y in range(m):
        matrix[x][y] = int(input())

vector = np.zeros(m)

for i in range(m):
    vector[i] = int(input())

k = int(input())

percentage_matrix = np.zeros((s, m), dtype=int)

for x in range(s):
    for y in range(m):
        percentage_matrix[x][y] = 100 * matrix[x][y] / vector[y]
        print(percentage_matrix[x][y])
    print()  
        


# matrix[x, :] = row x, matrix[:, y] = column y
print("Salesperson Summary:")
for x in range(s):
    avg_percentage = np.mean(percentage_matrix[x])
    best_product = np.argmax(percentage_matrix[x])
    print(f"Salesperson {x}: Average Achievement = {avg_percentage:.2f}%, "
          f"Best Product = {best_product}")

print("Product Summary:")
for y in range(m):
    avg_percentage = np.mean(percentage_matrix[:,y])
    best_product = np.argmax(percentage_matrix[:,y])
    print(f"Product {y}: Average Achievement = {avg_percentage:.2f}%, "
          f"Best Salesperson = {best_product}")

# axis = 0 → column-wise operation
# axis = 1 → row-wise operation

avg_percentages = np.mean(percentage_matrix, axis=1)
top_k_ids = np.argsort(avg_percentages)[::-1]
top_k_ids = top_k_ids[:k]

print(f"Top {k} Salesperson IDs: {top_k_ids}")


conditions = [
    percentage_matrix >= 90,
    (percentage_matrix >= 75) & (percentage_matrix < 90),
    (percentage_matrix >= 60) & (percentage_matrix < 75),
    percentage_matrix < 60
]
grades = ["Excellent", "Good", "Average", "Poor"]


# np.select(conditions, grades) — element-wise: for each cell, it checks the conditions in order and picks the matching label from grades.
grade_matrix = np.select(conditions, grades, default="Unknown")

# flattens the matrix, finds each distinct grade string, and counts how many times it occurs. Returns two parallel arrays: unique (the distinct labels) and counts (their counts).
unique, counts = np.unique(grade_matrix, return_counts=True)

# zips those two arrays into {grade: count} pairs.
grade_counts = dict(zip(unique, counts))


for grade in grades:
    print(f"{grade}: {grade_counts.get(grade, 0)}")
```

```python
# Problem 4
import numpy as np
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

north = np.array([120, 135, 128, 142, 150, 158, 162, 170, 168, 180, 185, 195])
south = np.array([110, 118, 125, 130, 138, 145, 150, 155, 160, 168, 172, 180])
central = np.array([100, 108, 115, 120, 130, 140, 148, 152, 158, 165, 170, 175])
```

```python
fig, ax = plt.subplots(2,2,figsize=(10, 6))
# 'fig' is the whole window, 'axs' is a 2D list containing our 4 chart areas.

# Add the main title to the very top of the entire figure
fig.suptitle('Company Branch Performance Analysis (2025)', fontsize=16)

ax[0, 0].plot(months, north, color='blue', linestyle='solid', label='North')
ax[0, 0].plot(months, south, color='green', linestyle='dashed', label='South')
ax[0, 0].plot(months, central, color='red', linestyle='dotted', label='Central')

# Add the labels, title, legend
ax[0, 0].set_title('Monthly Branch Profit (2025)')
ax[0, 0].set_xlabel('Month')
ax[0, 0].set_ylabel('Profit (in thousand dollars)')
ax[0, 0].legend() # This makes the key showing which line is which branch
ax[0, 0].grid(True)

branches = ['North', 'South', 'Central']
profits = [north[-1], south[-1], central[-1]]  
ax[0, 1].bar(branches, profits, color=['blue', 'green', 'red'])
ax[0, 1].set_title('December Profit Comparison')
ax[0, 1].set_ylabel('Profit (in thousand dollars)')
ax[0,1].set_xlabel('Branch')
ax[0,1].grid(axis='y')

ax[1,0].scatter(months, north, color='blue', marker='o', s = 60)
ax[1,0].set_title('North Branch Profit Scatter Plot')
ax[1,0].set_xlabel('Month')
ax[1,0].set_ylabel('Profit (in thousand dollars)')
ax[1,0].grid(True)

quarters = ['Q1', 'Q2', 'Q3', 'Q4']

# Calculate quarterly sums using array slicing. [0:3] means index 0, 1, 2 (Jan, Feb, Mar)
north_q = [north[0:3].sum(), north[3:6].sum(), north[6:9].sum(), north[9:12].sum()]
south_q = [south[0:3].sum(), south[3:6].sum(), south[6:9].sum(), south[9:12].sum()]
central_q = [central[0:3].sum(), central[3:6].sum(), central[6:9].sum(), central[9:12].sum()]

# To make a stacked bar chart, we use the "bottom" parameter. 
# South sits on top of North. Central sits on top of (North + South).
ax[1, 1].bar(quarters, north_q, label='North', color='blue')
ax[1, 1].bar(quarters, south_q, bottom=north_q, label='South', color='green')

# Calculate the starting height for the Central bars (North + South)
bottom_central = np.add(north_q, south_q) 
ax[1, 1].bar(quarters, central_q, bottom=bottom_central, label='Central', color='red')

ax[1, 1].set_title('Quarterly Branch Profit')
ax[1, 1].set_xlabel('Quarter')
ax[1, 1].set_ylabel('Profit (in thousand dollars)')
ax[1, 1].legend()
ax[1, 1].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
